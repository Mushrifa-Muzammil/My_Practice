```python
# 3. Body Mass Index (BMI)

# Take input from user
bmi = float(input("Enter the BMI Index: "))

# Check BMI category
if (bmi < 18.5):
    print("Underweight")
elif (18.5 <= bmi < 25):
    print("Normal weight")
elif (25 <= bmi < 30):
    print("Overweight")
else:
    print("Very Overweight")
```
